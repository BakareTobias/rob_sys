## ✏️ Task 1.0 - Rotations
In the previous notebook, you used a provided rotation matrix to transform a vector between coordinate frames. Now, you will construct the rotation matrices from individual rotations and use them to describe the orientation of a body.

The orientation of a robot can be described using successive elementary rotations.
- the rotation about the $x$-axis
- the rotation about the $y$-axis
- the rotation about the $z$-axis

Their corresponding rotation matrices are
\begin{equation*}
\boldsymbol{R}_{x}(\alpha) =
\begin{bmatrix}
 1 & 0 & 0 \\
0 & \cos(\alpha)& -\sin(\alpha)  \\
 0 & \sin(\alpha) & \cos(\alpha)  \\
\end{bmatrix},
\end{equation*}

\begin{equation*}
\boldsymbol{R}_{y}(\alpha) =
\begin{bmatrix}
 \cos(\alpha) & 0 & \sin(\alpha) \\
0 & 1& -0  \\
 -\sin(\alpha) & 0 & \cos(\alpha)  \\
\end{bmatrix},
\end{equation*}

\begin{equation*}
\boldsymbol{R}_{z}(\alpha) =
\begin{bmatrix}
 \cos(\alpha)& -\sin(\alpha) & 0 \\
\sin(\alpha) & \cos(\alpha)& 0  \\
 0 & 0 & 1  \\
\end{bmatrix}
\end{equation*}


---
### Task 1.1

Using the definitions above, complete the functions below to construct rotation matrices about the $x$-, $y$-, and $z$-axes.

```python
# Task 1.1

def Rx(alpha):


    # Convert degrees to radians
    a = np.radians(alpha)

    # Calculate sine and cosine
    ca = np.cos(a)
    sa = np.sin(a)

    # Produce rotation matrix about x-axis
    Rx = np.array([[1,0,0],[0,ca,-sa],[0,sa,ca]])

    return Rx


def Ry(alpha):


    # Convert degrees to radians
    a = np.radians(alpha)

    # Calculate sine and cosine
    ca = np.cos(a)
    sa = np.sin(a)

    # Produce rotation matrix about y-axis
    Ry = np.array([[ca,0,sa],[0,1,-0],[-sa,0,ca]])

    return Ry


def Rz(alpha):


    # Convert degrees to radians
    a = np.radians(alpha)

    # Calculate sine and cosine
    ca = np.cos(a)
    sa = np.sin(a)

    # Produce rotation matrix about z-axis
    Rz = np.array([[ca,-sa,0],[sa,ca,0],[0,0,1]])

    return Rz
```

## Rotations with Respect to the Fixed or Current Frame

When several rotations are combined, we must specify **which reference frame the rotations are performed with respect to**.

* **Rotation w.r.t. the fixed frame:** Each rotation is performed about the axes of a fixed reference frame (e.g., the world frame). These axes remain unchanged as the body rotates.

* **Rotation w.r.t. the current frame:** Each rotation is performed about the axes of the current (body-fixed) frame. After each rotation, the frame moves with the body, so the next rotation is performed about the updated axes.

### Combining Successive Rotations

When combining successive rotations, the order of matrix multiplication depends on whether the rotations are performed **w.r.t. the fixed frame or the current frame**.

Let $\mathbf{R}$ represent the current orientation of the body with respect to the fixed frame, and let $\mathbf{R}_{rotation}$ represent an additional rotation.

**1. Rotation w.r.t. the current frame**

The new rotation is applied from the **right** (post-multiplication):

$$
\mathbf{R}_{new}
=
\mathbf{R}\mathbf{R}_{rotation}.
$$

**2. Rotation w.r.t. the fixed frame**

The new rotation is applied from the **left** (pre-multiplication):

$$
\mathbf{R}_{new}
=
\mathbf{R}_{rotation}\mathbf{R}.
$$

The multiplication order matters because rotation matrices are generally **non-commutative**. Always identify whether each rotation is defined w.r.t. the **fixed frame** or the **current frame** before combining them.

## ✏️ Task 2.0 - Homogeneous Transformation: Combining Translations and Rotations

The **Homogeneous Transformation Matrix (HTM)** provides a compact representation of both **rotation** and **translation** between two coordinate frames. Homogeneous transformations can be **chained** through matrix multiplication and **reversed** by taking the inverse, making them particularly useful for representing rigid-body poses and transformations in robotics.

A homogeneous transformation matrix is defined as

$$
\boldsymbol{H} =
\begin{bmatrix}
\color{red}{\boldsymbol{R}} & \color{green}{\boldsymbol{p}}\\
\boldsymbol{0}_{1\times3} & 1
\end{bmatrix},
$$

where

$$
\color{red}{\boldsymbol{R}} \in \mathbb{R}^{3\times3}
$$

is the **rotation matrix**, describing the orientation, and

$$
\color{green}{\boldsymbol{p}} =
\begin{bmatrix}
p_x\\
p_y\\
p_z
\end{bmatrix}
\in \mathbb{R}^{3}
$$

is the **position vector**, describing the translation.

For example, if the pose of frame $\mathcal{F}_B$ relative to $\mathcal{F}_A$ is described by $\boldsymbol{H}^{A}_{B}$, and the pose of frame $\mathcal{F}_C$ relative to $\mathcal{F}_B$ is described by $\boldsymbol{H}^{B}_{C}$, then the pose of $\mathcal{F}_C$ relative to $\mathcal{F}_A$ is

$$
\boldsymbol{H}^{A}_{C}
=
\boldsymbol{H}^{A}_{B}\boldsymbol{H}^{B}_{C}.
$$

The transformation can also be reversed using the inverse:

$$
\boldsymbol{H}^{B}_{A}
=
\left(\boldsymbol{H}^{A}_{B}\right)^{-1}.
$$

To transform a **3D point** using a homogeneous transformation matrix, the point must first be represented using **homogeneous coordinates**. This is done by appending a $1$ to the position vector:

$$
\boldsymbol{p}
=
\begin{bmatrix}
p_x\\
p_y\\
p_z
\end{bmatrix}
\quad\longrightarrow\quad
\bar{\boldsymbol{p}}
=
\begin{bmatrix}
p_x\\
p_y\\
p_z\\
1
\end{bmatrix}.
$$

The resulting 4-vector is called the **homogeneous coordinate representation** of the point.

A point expressed in frame $\mathcal{F}_B$ can then be transformed into frame $\mathcal{F}_A$ using
$$
\bar{\boldsymbol{p}}^{A}
=
\boldsymbol{H}^{A}_{B}\,\bar{\boldsymbol{p}}^{B}.
$$

The first three elements of the resulting vector give the $x$, $y$, and $z$ coordinates of the point expressed in frame $\mathcal{F}_A$.

---
### Task 2.1
Complete the function below to construct a Homogeneous Transformation Matrix from a rotation matrix and a translation vector.


```python
# Task 2.1
def ht(r, t):
    """
    Construct a 4x4 Homogeneous Transformation Matrix (HTM).

    Inputs:
        r : numpy.ndarray
            A 3x3 rotation matrix describing the orientation.
        t : numpy.ndarray
            A translation vector with 3 components [x, y, z].

    Output:
        ht_out : numpy.ndarray
            A 4x4 homogeneous transformation matrix.
    """

    if r.shape != (3, 3):
        raise ValueError("Rotation matrix must be 3x3")

    if t.size != 3:
        raise ValueError("Translation vector must have 3 components")

    #t = t.reshape(3,)

    # Start with a 4x4 identity matrix
    ht_out = np.identity(4)

    # Add the rotation matrix
    ht_out[:3 ,:3] = r

    # Add the translation vector
    # Hint: ensure t has shape (3,) before assigning it
    ht_out[:3, 3] = t.reshape(-1)

    return ht_out
```

## ✏️ Task 3.0 - Localising an Object from a Moving Underwater Vehicle

Now, consider the underwater vehicle equipped with the same sensor used in **Task 2**, with the same relative pose between the sensor and vehicle.

The system contains three coordinate frames:

- $\mathcal{F}_W$: the world-fixed frame
- $\mathcal{F}_B$: the vehicle body frame
- $\mathcal{F}_S$: the sensor frame

Because the sensor is rigidly mounted to the vehicle, the transformation

$$
\mathbf{H}^{B}_{S}
$$

calculated in Task 2 remains constant.

However, as the vehicle moves, its pose relative to the world frame,

$$
\mathbf{H}^{W}_{B},
$$

changes.

A point measured by the sensor can be expressed in the world frame using the transformation chain

$$
\mathbf{H}^{W}_{S}
=
\mathbf{H}^{W}_{B}
\mathbf{H}^{B}_{S}.
$$

In this task, you will combine these transformations to determine the position of a stationary object in the world frame as the vehicle moves.

---
### Task 3.1

Complete the function below to determine the position of an object in the world frame.

The function is provided with:

- the position of the vehicle in the world frame,
- Three rotation angles describing the vehicle's orientation, corresponding to successive rotations about the current z-, y-, and x-axes, respectively.
- and the position of an object measured in the sensor frame.

Using the rotation and HTM functions developed in the previous tasks, calculate

$$
\mathbf{p}^{W}_{P}.
$$



```python
# Task 3.1

def localise_object(p_W_B, angle_x, angle_y, angle_z, p_S_P):
    """
    Determine the position of an object in the world frame.

    Inputs:
        p_W_B : numpy.ndarray
            Position of the vehicle in the world frame [x, y, z] (m).

        angle_x, angle_y, angle_z : float
            Vehicle rotation angles about the x-, y-, and z-axes (degrees).

        p_S_P : numpy.ndarray
            Position of the object measured in the sensor frame
            [x, y, z] (m).

    Output:
        p_W_P : numpy.ndarray
            Position of the object in the world frame [x, y, z] (m).
    """

    # Vehicle orientation relative to the world frame
    # Successive rotations about the current frame:
    # 1. Rotate about the current z-axis by angle_z
    # 2. Rotate about the updated y-axis by angle_y
    # 3. Rotate about the updated x-axis by angle_x

    R_WB = Rz(angle_z) @ Ry(angle_y) @ Rx(angle_x)

    # Vehicle pose relative to the world frame
    H_WB = ht(R_WB,p_W_B)

    # Sensor pose relative to the world frame
    H_WS = H_WB @ H_BS

    # Convert the sensor measurement into homogeneous coordinates
    p_S_P_homogeneous = ht(np.identity(3),p_S_P)

    # Transform the object into the world frame
    p_W_P_homogeneous = H_WS @ p_S_P_homogeneous

    # Return x, y, z coordinates
    return p_W_P_homogeneous[:3,3]
```

Using your rotation and HTM functions:

```python
# Task 4.3

# Rotation matrices
R_AB = Rx(90) @ Ry(0) @ Rz(45)
R_BC = Rz(-115)
R_CD = Ry(90)

# HTMs
H_AB = ht(R_AB,np.array([0,0,5]))
H_BC = ht(R_BC,np.array([15,0,0]))
H_CD = ht(R_CD,np.array([10,0,0]))

# Transform from base frame to camera frame
H_DA = np.linalg.inv(H_AB @ H_BC @ H_CD)

# Doughnut position in the base frame
p_D_doughnut = H_DA @ ht(np.eye(3),np.array([15,0,3]))
print(p_D_doughnut[:3,3])
```
