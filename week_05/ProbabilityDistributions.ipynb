Random processes can be described by distributions of probability.  In the simplest case for a coin flip, there are only two places for our probability to go, Heads or Tails.  This is known as a discrete case (as there are discrete values)

```python
# The Bernoulli function is used to track the probabilities of discrete outcomes

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli
from ipywidgets import interact
import ipywidgets as widgets

# Function to plot a Bernoulli distribution
def plot_bernoulli(p):
    x = [0, 1]
    pmf = bernoulli.pmf(x, p)
    
    plt.figure(figsize=(8, 5))
    plt.bar(x, pmf, label=f'p = {p}', color='blue', zorder=2)
    plt.xticks(x, ['A','B'])
    plt.xlabel('X')
    plt.ylabel('Probability')
    plt.ylim(0,1.1)
    # plt.legend()
    plt.title('Bernoulli Distribution')
    plt.grid(False)
    plt.show()

# Create an interactive slider for the probability parameter 'p'
p_slider = widgets.FloatSlider(value=0.5, min=0.0, max=1.0, step=0.1, description='p (Success Probability)')

# Create the interactive plot
interact(plot_bernoulli, p=p_slider)
```

```python
# The Poisson distribution (a top tier distribution in the eyes of Dr West), is used for the expected number of discrete events.  Originally, it was looking at the probability of how many soliders got kicked by a horse.
# A rate lambda is the average rate (for example cars passing in an hour), but of course you can not have fractions of a car, so it takes continuous variable (rate) and produces probabilities for integer values.
# At large values of lambda, it's shape approximates a normal distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from ipywidgets import interact
import ipywidgets as widgets

# Function to plot a Poisson distribution
def plot_poisson(lambda_param):
    x = np.arange(0, 20)  # Adjust the range based on your specific needs
    pmf = poisson.pmf(x, lambda_param)
    
    plt.figure(figsize=(10, 5))
    plt.bar(x, pmf, color='blue', label=f'λ = {lambda_param}', zorder=2)
    plt.xticks(x)
    plt.xlabel('X (Number of Events)')
    plt.ylabel('Probability')
    plt.ylim(0,0.35)
    # plt.legend()
    plt.title('Poisson Distribution')
    plt.grid(False)
    plt.show()

# Create an interactive slider for the λ (lambda) parameter
lambda_slider = widgets.FloatSlider(value=2.0, min=0.1, max=10.0, step=0.1, description='λ (Rate)')

# Create the interactive plot
interact(plot_poisson, lambda_param=lambda_slider)
```

```python
# The Uniform distribution is the most "naive" of distributions, that is we give all outcomes equal probability.
# This is commonly used when you have no prior knowledge of a system, or is it genuinely uniform!

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import ipywidgets as widgets

# Function to plot a uniform distribution
def plot_uniform_distribution(start, end, num_samples):
    x = np.linspace(start, end, 1000)
    y = np.where((x >= start) & (x <= end), 1 / (end - start), 0)

    # Sample points from the Uniform distribution
    samples = np.random.uniform(start, end, num_samples)
    
    plt.figure(figsize=(10, 5))
    plt.plot(x, y, label=f'Start = {start}, End = {end}', color='blue')

  
    # Add vertical lines at the start and end, stopping at the same height
    plt.vlines(x=[start, end], ymin=0, ymax=1 / (end - start), colors=['blue', 'blue'])
   
    # Create a scatter plot of the sampled points
    plt.scatter(samples, ([0.5] * num_samples), color='red', marker='o', s=50, label='Samples', alpha=0.2)

    plt.xlabel('X')
    plt.ylabel('Probability Density')
    # plt.legend()
    plt.title('Uniform Distribution')
    plt.xlim(-5,5)
    plt.ylim(0,1.0)
    plt.grid(True)
    plt.show()

# Create interactive sliders
start_slider = widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.1, description='Start')
end_slider = widgets.FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='End')
num_samples_slider = widgets.IntSlider(value=0, min=0, max=100, step=10, description='Num Samples')

# Create the interactive plot
interact(plot_uniform_distribution, start=start_slider, end=end_slider, num_samples=num_samples_slider)
```

```python
# The Beta distribution is a continuous cousin to the Bernoulli distribution.
# It can be used to track how biased a coin is for example, given two values alpha and beta which describe its shape.

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta
from ipywidgets import interact
import ipywidgets as widgets

# Function to plot a Beta distribution
def plot_beta(alpha, beta_param, num_samples):
    x = np.linspace(0, 1, 1000)
    pdf = beta.pdf(x, alpha, beta_param)


    
    plt.figure(figsize=(10, 5))
    plt.plot(x, pdf, color='blue', label=f'α = {alpha}, β = {beta_param}')
    # Create a scatter plot of the sampled points with transparency
    # Sample points from the Beta distribution
    samples = np.random.beta(alpha, beta_param, num_samples)
    plt.scatter(samples, ([0.5] * num_samples), color='red', marker='o', s=50, label='Samples', alpha=0.2)

    plt.xlabel('X')
    plt.ylabel('Probability Density')
    # plt.legend()
    plt.title('Beta Distribution')
    plt.grid(True)
    plt.show()

# Create interactive sliders for the alpha and beta parameters
alpha_slider = widgets.FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description='α (Alpha)')
beta_slider = widgets.FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description='β (Beta)')
num_samples_slider = widgets.IntSlider(value=0, min=0, max=100, step=10, description='Num Samples')


# Create the interactive plot
interact(plot_beta, alpha=alpha_slider, beta_param=beta_slider, num_samples=num_samples_slider)
```

```python
# The normal distribution, the most important of all distributions in probability and statistics.
# Due to the central limit theorem (https://youtu.be/YAlJCEDH2uY?si=I8P-FCN8gcCqrE76 and https://youtu.be/zeJD6dqJ5lo?si=pSgVJMU0QAp3fGsl),
# all distributions tend to this one.
# Furthermore, they are super mathematically convenient, for example, combining normal distributions is mathematically trivial (https://youtu.be/d_qvLDhkg00?si=JW5mOlPeiBRqLjon)
# Dr West may have a soft spot for the Poisson distribution, but the undisputed king is the normal distribution.


import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import ipywidgets as widgets

# Function to plot a normal distribution
def plot_normal_distribution(mean, variance, num_samples):
    x = np.linspace(-10, 10, 1000)
    y = 1 / np.sqrt(2 * np.pi * variance) * np.exp(-0.5 * ((x - mean) ** 2) / variance)
    
    plt.figure(figsize=(10, 5))
    plt.plot(x, y, label=f'Mean = {mean}, Variance = {variance}', color='blue')
    # Sample points from the Gaussian distribution
    samples = np.random.normal(mean, variance, num_samples)
    plt.scatter(samples, ([0.1] * num_samples), color='red', marker='o', s=50, label='Samples', alpha=0.2)
    plt.xlabel('X')
    plt.ylabel('Probability Density')
    # plt.legend()
    plt.title('Normal Distribution')
    plt.ylim(0,0.4)
    plt.grid(False)
    plt.show()

# Create interactive sliders
mean_slider = widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.1, description='Mean')
variance_slider = widgets.FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='Variance')
num_samples_slider = widgets.IntSlider(value=0, min=0, max=100, step=10, description='Num Samples')


# Create the interactive plot
interact(plot_normal_distribution, mean=mean_slider, variance=variance_slider, num_samples=num_samples_slider)
```

```python
# When we have noisy data, you can bet that it will be normally distributed
# In reality we have our data measurement with the addition of some noise, D + e
# When we perform analysis of data, we need to remember that we are trying to uncover what is D,
# whilst accounting for e.

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import ipywidgets as widgets

# Function to perform linear regression with noisy data
def linear_regression(variance, num_points):
    # Generate random data
    x = np.linspace(0, 10, num_points)
    true_slope = 2
    true_intercept = 1
    y_true = true_slope * x + true_intercept
    noise = np.random.normal(0, variance, num_points)
    y = y_true + noise
    
    # Perform linear regression
    # regression_coeff = np.polyfit(x, y, 1)
    # regression_line = np.polyval(regression_coeff, x)
    
    plt.figure(figsize=(10, 5))
    plt.scatter(x, y, label='Noisy Data', color='blue')
    # plt.plot(x, regression_line, label=f'Regression Line (Slope: {regression_coeff[0]:.2f}, Intercept: {regression_coeff[1]:.2f})', color='red')
    plt.plot(x, y_true, label=f'True Line (Slope: {true_slope}, Intercept: {true_intercept})', linestyle='--', color='green')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.ylim(-5, 30)
    # plt.legend()
    plt.title('Noisy Data')
    plt.grid(True)
    plt.show()

# Create interactive sliders for noise variance and number of data points
variance_slider = widgets.FloatSlider(value=0.0, min=0.0, max=5.0, step=0.1, description='Noise Variance')
num_points_slider = widgets.IntSlider(value=30, min=10, max=100, step=10, description='Number of Data Points')

# Create the interactive plot
interact(linear_regression, variance=variance_slider, num_points=num_points_slider)
```
