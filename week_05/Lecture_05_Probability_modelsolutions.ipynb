## ✏️ Task 4.0 - Bayes Rule: Part 1

As we saw with conditional probabilities, the more information we have, the better our understanding of the system.  Bayes Rule allows us to update our belief about a system when we encounter new information.  Like good engineers and scientists, our conclusions can change given new data.  Bayes Rule is typically written,

\begin{equation*}
p(H|E) = \dfrac{p(E|H) \times p(H)}{p(E)}
\end{equation*}

where $H$ is a "hypothesis" and $E$ is evidence.  By writing it more explicitly, it begins to reveal itself,

\begin{equation*}
p(H \cap E|E) = \dfrac{p(H \cap E|H) \times p(H)}{p(E)}
\end{equation*}

It is a method of taking known probabilities and infering information about the opposite question.  A classic example would be asking boys and girls in a school class if they think robots are cool.  From this data, we can use Bayes Rule to find the probability they are either a girl or a boy, given they like robots.  We can infer a "reverse" property of something based on the known property.  Remember is this just an estimation!


---
### Task 4.1

You are at a local school in Manchester performing outreach activities with some robots (for the school children - the robots are already very invested in STEM). The students are between 7-8 years old.

With a simple show of hands, you ask the question "Are robots cool? Yes/No".

The class contains 33 students, of whom 15 are boys and 18 are girls.

When asked the question, 10 boys and 9 girls raise their hands to indicate that they think robots are cool.

You then ask yourself: **What is the probability that a student is a girl, given that they think robots are cool?**

Before applying Bayes' rule, identify the following probabilities from the scenario:

1. The prior probability that a randomly selected student is a girl, $P(G)$.
2. The probability that a girl thinks robots are cool, $P(R|G)$.
3. The overall probability that a randomly selected student thinks robots are cool, $P(R)$.

Where $G$ represents the event that a student is a girl and $R$ represents the event that a student thinks robots are cool.

Using these probabilities, calculate $P(G|R)$ using Bayes' rule.


Answer:

First, identify the relevant probabilities.

There are 18 girls in a class of 33 students, therefore:

$$
P(G)=\frac{18}{33}
$$

Of the 18 girls, 9 think robots are cool:

$$
P(R|G)=\frac{9}{18}=0.5
$$

In total, 19 students think robots are cool:

$$
P(R)=\frac{10+9}{33}=\frac{19}{33}
$$

Now apply Bayes' rule:

$$
P(G|R)=\frac{P(R|G)\times P(G)}{P(R)}
$$

$$
P(G|R)=\frac{(9/18)\times(18/33)}{19/33}
=\frac{9}{19}\approx 0.474
$$

Therefore, the probability that a randomly selected student who thinks robots are cool is a girl is approximately **47.4%**.



*Sadly, gender stereotypes in STEM (Science, Technology, Engineering and Mathematics) are already forming at an early age<sup>[1]</sup>.

[1] - F. Law _et al._, Children's Gender Stereotypes in STEM Following a One-Shot Growth Mindset Intervention in a Science Museum, Frontiers of Psychology, (2021) https://doi.org/10.3389/fpsyg.2021.641695


```python

# Task 4.1

# Extract the information from the scenario
total_pop = 33
n_girl = 18
n_boy = 15

n_girl_robots = 9
n_boy_robots = 10

# Calculate prior probability and likelihood
p_g = n_girl / total_pop
p_r_given_g = n_girl_robots / n_girl

# Marginal probability
p_r = (n_girl_robots + n_boy_robots) / total_pop

# Apply Bayes' rule
p_g_given_r = (p_r_given_g * p_g) / p_r

print(f'Probability that a student who thinks robots are cool is a girl: {p_g_given_r:.3f}')
```

## ✏️ Task 5.0 - Bayes Rule: Part 2

Let us concentrate on the updating belief part of Bayes Rule.  The current belief (previous posterior) becomes the prior, and a new posterior is calculated using new evidence.

\begin{equation*}
\text{Posterior} = \dfrac{\text{Likelihood} \times \text{Prior}}{\text{Marginal}} = \dfrac{\text{Likelihood} \times \text{Previous Posterior}}{\text{Marginal}}
\end{equation*}

This iterative updating will be explored in the next few tasks.

### Task 5.1

A mobile robot can be in one of three possible positions: $A$, $B$ or $C$.

Using dead reckoning from its wheel encoders, the robot estimates that:

- There is a 20% chance that it is at position $A$.
- There is a 50% chance that it is at position $B$.
- There is a 30% chance that it is at position $C$.

The robot then receives data from an algorithm which attempts to match LiDAR scans to known locations.

The scan-matching algorithm is not always correct.

From previous testing, engineers know that:

- When the robot is actually at position $A$, the algorithm incorrectly reports position $B$ in 30 out of every 100 tests.
- When the robot is actually at position $B$, the algorithm correctly reports position $B$ in 80 out of every 100 tests.
- When the robot is actually at position $C$, the algorithm incorrectly reports position $B$ in 20 out of every 100 tests.

During the current measurement, the algorithm reports that the robot is at position $B$.

Before applying Bayes' rule, identify:

1. The prior probabilities $P(A)$, $P(B)$ and $P(C)$.
2. The likelihoods $P(S_B|A)$, $P(S_B|B)$ and $P(S_B|C)$, where $S_B$ means that the sensor reports position $B$.

Using these values, calculate the probability that the robot is actually at position $B$, given that the sensor reports $B$.

Calculate $P(B|S_B)$.

Answer:


```python
# Task 5.1

# Prior belief (probabilities)
P_A = 0.2
P_B = 0.5
P_C = 0.3

# Likelihoods of data (sensor says B given position)
P_SB_given_A = 0.3
P_SB_given_B = 0.8
P_SB_given_C = 0.2

# Marginal probability that the sensor would report "B" for any situation
P_SB = P_SB_given_A * P_A + P_SB_given_B * P_B + P_SB_given_C * P_C

# Apply Bayes Rule
P_B_given_SB = (P_SB_given_B * P_B) / P_SB

print(f'P(Position B | Sensor says B) =  {P_B_given_SB:.4f}')
```

### Task 5.6 - STRETCH

Write a Python function which can update the belief in position $B$, given a list of observations, e.g. ``['B', 'A', 'B', 'C']``.

The final set of likelihoods required are for the sensor reporting $C$:

\begin{align*}
P(S_{C} | A) &= 0.0\\
P(S_{C} | B) &= 0.1\\
P(S_{C} | C) &= 0.5\\
\end{align*}

Hint, you could use Python dictionaries to lookup likelihood values.

```python
# Task 5.6 - STRETCH

def update_belief_B_after_observations(observations, P_A=1/3, P_B=1/3, P_C=1/3):
    # Default values for P_A = P_B = P_C = 1/3 (equal probability)
    
    # Likelihoods for sensor readings
    likelihoods = {
        'A': {'A': 0.7, 'B': 0.1, 'C': 0.3},  # P(sensor says A | robot at A, B, C)
        'B': {'A': 0.3, 'B': 0.8, 'C': 0.2},  # P(sensor says B | robot at A, B, C)
        'C': {'A': 0.0, 'B': 0.1, 'C': 0.5},  # P(sensor says C | robot at A, B, C)
    }
    
    # Iterate over each observation and update beliefs
    # If you were feeling snazzy, this could be done recursively
    for observation in observations:
        # Step 1: Compute the total probability of the observation (normalizing constant)
        P_sensor = (
            likelihoods[observation]['A'] * P_A +
            likelihoods[observation]['B'] * P_B +
            likelihoods[observation]['C'] * P_C
        )
        
        # Step 2: Update the posterior beliefs using Bayes' Rule
        P_A = (likelihoods[observation]['A'] * P_A) / P_sensor
        P_B = (likelihoods[observation]['B'] * P_B) / P_sensor
        P_C = (likelihoods[observation]['C'] * P_C) / P_sensor

    # Return the posterior belief in B
    return P_B

# Test using scenario from 5.1-5.4
observations = ['B', 'A']
updated_belief_in_B = update_belief_B_after_observations(observations, P_A=0.2, P_B=0.5, P_C=0.3)

print(f"Updated belief in B after observations: {updated_belief_in_B:.4f}")
```

P(A|D)=\frac{P(D|A)\times P(A)}{P(D)}
$$

First, calculate the marginal probability of a positive camera detection:

$$
P(D)=P(D|A)P(A)+P(D|\neg A)P(\neg A)
$$

$$
P(D)=(0.91\times0.35)+(0.14\times0.65)
$$

$$
P(D)=0.4095
$$

Now apply Bayes' rule:

$$
P(A|D)=\frac{0.91\times0.35}{0.4095}
$$

$$
P(A|D)\approx0.7778
$$

Therefore, after receiving the camera measurement, the robot has approximately 77.78% confidence that the opening is actually accessible.

### Task 6.4

The robot now uses its LiDAR sensor to further analyse the geometry of the same opening.

The LiDAR system also reports that the opening is accessible.

From previous testing, engineers know that:

- When an opening is actually accessible, the LiDAR correctly identifies it as accessible 85% of the time.
- When an opening is not accessible, the LiDAR incorrectly classifies it as accessible 20% of the time.

Assume that the camera and LiDAR measurements are conditionally independent, given whether the opening is actually accessible.

**Using the posterior probability from Task 6.2 as the new prior**, apply Bayes' rule again to calculate the probability that the opening is accessible after both sensor measurements.

Calculate $P(A|D,L)$.

Answer:

First, identify the likelihoods from the scenario:

$$
P(L|A)=0.85
$$

$$
P(L|A')=0.20
$$

From Task 6.2, the previous posterior was:

$$
P(A|D)\approx0.7778
$$

Therefore:

$$
P(A'|D)\approx1-0.7778=0.2222
$$

Assuming conditional independence of the two sensors, we can apply Bayes' rule again:

$$
P(A|D,L)=\frac{P(L|A)\times P(A|D)}{P(L|D)}
$$

Here, the denominator is the probability of the LiDAR reporting an accessible opening, given the earlier camera detection:

$$
P(L|D)=P(L|A)P(A|D)+P(L|A')P(A'|D)
$$

Hence:

$$
P(A|D,L)=\frac{P(L|A)P(A|D)}{P(L|A)P(A|D)+P(L|A')P(A'|D)}
$$

Substituting the values:

$$
P(A|D,L)=\frac{0.85\times0.7778}{(0.85\times0.7778)+(0.20\times0.2222)}
$$

$$
P(A|D,L)\approx0.9370
$$

Therefore, after combining the camera and LiDAR measurements, the robot has approximately 93.70% confidence that the opening is accessible.
