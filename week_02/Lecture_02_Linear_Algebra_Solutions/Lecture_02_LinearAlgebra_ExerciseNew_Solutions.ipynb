---
### Task 1.1
Represent the position of the robot, the goal and the centre of the obstacles as numpy arrays.

````python
# Task 1.1
robot_position = np.array([1.0,1.0])
goal_position = np.array([3.5,3.5])
obstacle_1 = np.array([2.5,2.5])
obstacle_2 = np.array([3.0,1.7])
obstacle_3 = np.array([3.0,1.7])
````

---
### Task 1.2
Calculate the displacement vector from the robot’s current position to the goal, the Euclidean distance between the robot and the goal, and the unit vector that points from the robot towards the goal.

````python
# Task 1.2
displacement = goal_position - robot_position
distance_to_goal = np.linalg.norm(displacement)
direction_to_goal = displacement / distance_to_goal

print("Displacement:", displacement)
print("Distance to goal:", distance_to_goal)
print("Direction to goal:", direction_to_goal)
````

---
### Task 1.3
Given that the robot’s current direction of travel is  $\begin{bmatrix}0.7 & 0.4 \\ \end{bmatrix}^{T}$, determine how closely it is aligned with the direction towards the goal and interpret this result physically.

````python
# Task 1.3
direction_of_travel = np.array([0.7, 0.4])
direction_of_travel = (
    direction_of_travel /
    np.linalg.norm(direction_of_travel)
)

alignment = np.dot(
    direction_of_travel,
    direction_to_goal
)

alignment  = np.arccos(alignment)
print("Direction of travel:", direction_of_travel)
print("Alignment:", np.rad2deg(alignment))
````

**1. Physical Interpretation of the Alignment Value:**


The dot product between two vector is equal to the cosine of the angle between them. Therefore the direction of the goal and the direction of the robot is around 15.25 degrees apart

### Task 1.4
Calculate the Euclidean distance between the robot and each obstacle. Using these distances, determine which obstacle is closest to the robot.

````python
# Task 1.4

distance_to_obstacle_1 = np.linalg.norm(
    obstacle_1 - robot_position
)

distance_to_obstacle_2 = np.linalg.norm(
    obstacle_2 - robot_position
)

distance_to_obstacle_3 = np.linalg.norm(
    obstacle_3 - robot_position
)

obstacle_distances = np.array([
    distance_to_obstacle_1,
    distance_to_obstacle_2,
    distance_to_obstacle_3
])
closest_obstacle = np.argmin(obstacle_distances) + 1

print("Distance to obstacle 1:", distance_to_obstacle_1)
print("Distance to obstacle 2:", distance_to_obstacle_2)
print("Distance to obstacle 3:", distance_to_obstacle_3)

print("Closest obstacle:", closest_obstacle)
````

## ✏️ Task 2.0 - Coordinate Transformations:
<figure>
    <img src="graphics/task_2.png" width="600">
    <figcaption><b>Figure 2:</b> Mobile underwater vehicle</figcaption>
</figure>
Figure 2 shows a mobile underwater vehicle described using two coordinate frames:

- the **world frame** $\mathcal{F}_W$, which is fixed to the environment; and
- the **body frame** $\mathcal{F}_B$, which is attached to the vehicle and moves with it.

A physical quantity such as velocity can be expressed relative to either coordinate frame.

The rotation matrix below describes the orientation of the body frame relative to the
world frame:

$$
\mathbf{R}_{B}^{W} =
\begin{bmatrix}
0.7912 & -0.6017 & -0.1092 \\
0.5540 & 0.7809 & -0.2884 \\
0.2588 & 0.1677 & 0.9513
\end{bmatrix}.
$$

The velocity of the vehicle, expressed in its body-fixed coordinate frame, is given by

$$
\mathbf{v}^{B} =
\begin{bmatrix}
1.2 \\
0.3 \\
-0.1
\end{bmatrix}
\text{ m/s}.
$$

This means that the velocity components are currently described along the axes of
the vehicle's body frame $\mathcal{F}_B$.

To express the **same physical velocity** in the world frame $\mathcal{F}_W$, we use
the rotation matrix:
$$
\mathbf{v}^{W} = \mathbf{R}^{W}_{B}\mathbf{v}^{B}.
$$


---
### Task 2.1
Represent the rotation matrix and the vehicle’s body-frame velocity as NumPy arrays.

````python
# Task 2.1


R_WB = np.array([
    [0.7912, -0.6017, -0.1092],
    [0.5540,  0.7809, -0.2884],
    [0.2588,  0.1677,  0.9513]
])

v_B = np.array([
    1.2,
    0.3,
    -0.1
])
````

---
### Task 2.2
A rotation matrix is orthogonal, meaning
$$
\mathbf{R}^{T}\mathbf{R} = \mathbf{I}.
$$
Therefore, calculate $
\mathbf{R}^{T}\mathbf{R}
$ and verify that the result is equal to the identity matrix.

````python
# Task 2.2

orthogonality_check = R_WB.T @ R_WB

is_orthogonal = np.allclose(
    orthogonality_check,
    np.eye(3),
    atol=1e-4
)

print(orthogonality_check)
print("Is orthogonal:", is_orthogonal)
````

---
### Task 2.3
Use the rotation matrix to express the vehicle's body-frame velocity in the world frame

````python
# Task 2.3

# vehicles velocity expressed in the world frame

v_W = R_WB @ v_B

print("World-frame velocity:", v_W)
````

---
### Task 2.4
Additionally, a rotation matrix satisfies

$$ \mathbf{R}^{-1}=\mathbf{R}^{T}. $$

Use the transpose of the rotation matrix to transform the world-frame velocity back into the body frame:

$$ \mathbf{v}^{B} = \left(\mathbf{R}^{W}_{B}\right)^T\mathbf{v}^{W}. $$

Verify that the recovered velocity is the same as the original velocity.

````python
# Task 2.4

v_B_recovered = R_WB.T @ v_W

is_same = np.allclose(
    v_B_recovered,
    v_B,
    atol=1e-3
)

print("Recovered body-frame velocity:", v_B_recovered)
print("Matches original:", is_same)
````

---
### Task 2.5
The vehicle now encounters a wave. Assume the wave produces a small constant acceleration on the vehicle in the world frame:

$$ \mathbf{a}^{W}_{wave} = \begin{bmatrix} 0.10\\ -0.05\\ 0.02 \end{bmatrix} \text{ m/s}^2. $$

Its velocity therefore evolves according to

$$ \mathbf{v}^{W}(t+\delta t) = \mathbf{v}^{W}(t) +  \mathbf{a}^{W}_{wave}\delta t.$$

Therefore, calculate the velocity vehicle in the **world frame** after:
- 1 $s$
- 2 $s$
- 3 $s$

````python
# Task 2.5

a_wave_W = np.array([
    0.10,
    -0.05,
    0.02
])

time_intervals = np.array([
    1.0,
    2.0,
    3.0
])

velocities_W = []

for dt in time_intervals:

    v_new_W = v_W + a_wave_W * dt

    velocities_W.append(v_new_W)

    print(
        f"Velocity after {dt:.0f} s: "
        f"{v_new_W} m/s"
    )
````

---
### Task 2.6
For each time interval, calculate the new body-frame velocity and its magnitude.

````python
# Task 2.6
for dt, v_new_W in zip(time_intervals, velocities_W):

    v_new_B = R_WB.T @ v_new_W

    speed = np.linalg.norm(v_new_B)

    print(f"After {dt:.0f} s:")
    print("Body-frame velocity:", v_new_B)
    print("Speed:", speed, "m/s")
````

## ✏️ Task 3.0 - Linear Mappings and the Jacobian — Pan-Tilt Camera

<figure>

Figure 3 shows a camera mounted on a 2-DOF pan-tilt mechanism.

The mechanism is controlled by the two joint angles:

- $q_1$: pan angle, corresponding to rotation about the $z$-axis
- $q_2$: tilt angle, corresponding to rotation about the $x$-axis

The tilt joint is positioned a fixed height $h$ above the pan joint, and the camera is located a distance $L$ from the tilt joint along the y-axis.


The side and top views in Figure 3 show the geometry associated with the tilt and pan rotations respectively.

For this task, use

$$
L = 0.5 \text{ m},
\qquad
h = 0.3 \text{ m}.
$$




---
### Task 3.1

Using the geometry in Figure 3, derive expressions for the Cartesian position of the camera

$$
\mathbf{x}
=
\begin{bmatrix}
x\\
y\\
z
\end{bmatrix}
$$

as a function of the joint angles $q_1$ and $q_2$.

Use your expressions to complete the implementation of the function below.

````python
# Task 3.1
def camera_position(q1,q2):
    L = 0.5
    h = 0.3
    x =  L * np.cos(q2) * np.cos(q1)
    y =  L * np.cos(q2) * np.sin(q1)
    z =  h + L * np.sin(q2)
    return np.array([x,y,z])
````

---
### Task 3.2

Verify your implementation by evaluating the camera position for

$$
q_1 = 30^\circ,
\qquad
q_2 = 20^\circ.
$$

Remember that NumPy trigonometric functions use radians.

````python
# Task 3.2

q1 = np.deg2rad(30)
q2 = np.deg2rad(20)
x = camera_position(q1, q2)

print("Camera position:", x, "m")
````

---
### Task 3.3

The linear velocity of the camera is related to the joint velocities by

$$
\dot{\mathbf{x}}
=
\mathbf{J}(\mathbf{q})
\dot{\mathbf{q}},
$$

where

$$
\dot{\mathbf{q}}
=
\begin{bmatrix}
\dot{q}_1\\
\dot{q}_2
\end{bmatrix}.
$$

The position Jacobian is obtained by differentiating the camera position with respect to the joint variables:

$$
\mathbf{J}(\mathbf{q})
=
\frac{\partial \mathbf{x}}
{\partial \mathbf{q}}.
$$

Using the position equations derived in Task 3.1, derive the Jacobian and complete the implementation below.

````python
# Task 3.3

def camera_jacobian(q1, q2):

    L = 0.5

    J = np.array([
        [
            -L * np.sin(q1) * np.cos(q2),
            -L * np.cos(q1) * np.sin(q2)
        ],
        [
            L  * np.cos(q1) * np.cos(q2),
            -L * np.sin(q1) * np.sin(q2)
        ],   
        [
            0.0,
            L * np.cos(q2)
        ]
    ])

    return J
````

---
### Task 3.4

Evaluate the camera Jacobian for

$$
q_1 = 30^\circ,
\qquad
q_2 = 20^\circ.
$$


````python
# Task 3.4

q1 = np.deg2rad(30)
q2 = np.deg2rad(20)

J = camera_jacobian(q1, q2)

print("Jacobian:")
print(J)
````

---
### Task 3.5

At the same configuration, suppose the joint velocities are

$$
\dot{\mathbf{q}}
=
\begin{bmatrix}
0.4\\
-0.2
\end{bmatrix}
\text{ rad/s}.
$$

Calculate the resulting linear velocity of the camera using

$$

=
\mathbf{J}\dot{\mathbf{q}}.
$$

````python
# Task 3.5

q_dot = np.array([
    0.4,
    -0.2
])

x_dot = J @ q_dot

print("Camera velocity:", x_dot, "m/s")
````

## ✏️ Task 4.0 - Solving Linear Systems and the Pseudoinverse — Pan-Tilt Camera

In Task 3, the Jacobian was used to map joint velocities to the linear velocity of the camera:

$$
\dot{\mathbf{x}}
=
\mathbf{J}\dot{\mathbf{q}}.
$$

We now consider the inverse problem:

> Given a desired Cartesian velocity of the camera, determine the joint velocities required to produce it.

This is known as **inverse differential kinematics**.

---
### Task 4.1

Before calculating the required joint velocities, inspect the Jacobian derived in Task 3.

1. What are the dimensions of $\mathbf{J}$?
2. Based on these dimensions, is the inverse differential kinematics problem overdetermined, underdetermined or square?
3. Does the standard matrix inverse $\mathbf{J}^{-1}$ exist?
4. Why might the pseudoinverse be useful for this system?

#### Solution

The Jacobian has dimensions

$$
\mathbf{J}\in\mathbb{R}^{3\times2}.
$$

It is therefore a **tall matrix**.

There are three Cartesian velocity components,

$$
\dot{\mathbf{x}}
=
\begin{bmatrix}
\dot{x}\\
\dot{y}\\
\dot{z}
\end{bmatrix},
$$

but only two joint velocities,

$$
\dot{\mathbf{q}}
=
\begin{bmatrix}
\dot{q}_1\\
\dot{q}_2
\end{bmatrix}.
$$

The inverse differential kinematics problem is therefore **overdetermined**.

Because $\mathbf{J}$ is not square, the standard inverse

$$
\mathbf{J}^{-1}
$$

does not exist.

The pseudoinverse can instead be used to calculate the joint velocities that provide the least-squares approximation to a desired Cartesian camera velocity.

---
### Task 4.2

For a desired camera linear velocity

$$
\dot{\mathbf{x}}_d,
$$

the required joint velocities can be estimated using the pseudoinverse of the Jacobian:

$$
\dot{\mathbf{q}}
=
\mathbf{J}^{+}\dot{\mathbf{x}}_d.
$$

Consider the camera configuration

$$
q_1=30^\circ,
\qquad
q_2=20^\circ,
$$

and suppose we want the camera to move with the desired velocity

$$
\dot{\mathbf{x}}_d
=
\begin{bmatrix}
-0.07\\
-0.06\\
0.05
\end{bmatrix}
\text{ m/s}.
$$

Calculate the required joint velocities to achieve this desired camera velocity

<sub><sup>Hint: NumPy provides the pseudoinverse using `np.linalg.pinv()`.</sup></sub>

````python
# Task 4.2 - Solution

q1 = np.deg2rad(30)
q2 = np.deg2rad(20)

x_dot_desired = np.array([
    -0.07,
    -0.06,
     0.05
])

J = camera_jacobian(q1, q2)

# Calculate the pseudoinverse of the Jacobian
J_pinv = np.linalg.pinv(J)

# Calculate the required joint velocities
q_dot = J_pinv @ x_dot_desired

print("Required joint velocities:")
print(q_dot)
````

---
### Task 4.3

Use the joint velocities calculated in Task 4.2 to compute the resulting camera velocity:

$$
\dot{\mathbf{x}}_{\mathrm{achieved}}
=
\mathbf{J}\dot{\mathbf{q}}.
$$

Compare the achieved velocity with the desired velocity $\dot{\mathbf{x}}_d$.

1. Are the two velocities exactly equal?
2. Calculate any discrepancy between them and discuss


````python
# Task 4.3 - Solution

x_dot_achieved = J @ q_dot

print("Desired Cartesian velocity:")
print(x_dot_desired)

print("\nAchieved Cartesian velocity:")
print(x_dot_achieved)

# Compare the achieved velocity with the desired velocity
discrepancy = x_dot_achieved - x_dot_desired

print("\nDiscrepancy:")
print(discrepancy)
````

The achieved camera velocity is very close to, but not exactly equal to, the desired velocity.


The Jacobian is a overdetermined and as a result not every arbitrary three-dimensional velocity can be produced exactly.

The pseudoinverse returns the joint velocities that minimise the least-squares error

$$
\left\|
\mathbf{J}\dot{\mathbf{q}}
-
\dot{\mathbf{x}}_d
\right\|.
$$

### Task 4.4

A **singular configuration** occurs when the Jacobian loses rank.

Recall the Jacobian derived in Task 3:

Answer the following:

1. What is the maximum possible rank of $\mathbf{J}$?
2. What common term appears in the first column?
3. For what values of $q_2$ does this term become zero?
4. What happens to the first column at these configurations?
5. What therefore happens to the rank of the Jacobian?
6. What does this mean physically for the pan motion of the camera?

#### Solution

The Jacobian has two columns, so its maximum possible rank is 2

The first column is

$$
\begin{bmatrix}
-L\sin(q_1)\cos(q_2)\\
L\cos(q_1)\cos(q_2)\\
0
\end{bmatrix}.
$$

The common term is therefore

$$
\boxed{\cos(q_2)}.
$$

This term becomes zero when

$$
\cos(q_2)=0.
$$

This occurs at

$$
{q_2=\pm90^\circ}.
$$

At these configurations, the first column becomes

$$
\begin{bmatrix}
0\\
0\\
0
\end{bmatrix}.
$$

The Jacobian therefore loses rank:

$$
{\operatorname{rank}(\mathbf{J})=1}.
$$

From the geometry in Figure 3,

$$
r=L\cos(q_2).
$$

At

$$
q_2=\pm90^\circ,
$$

we have

$$
r=0.
$$

The camera therefore lies directly above or below the pan axis.

Rotating the pan joint $q_1$ no longer changes the Cartesian position of the camera, which explains why the first column of the Jacobian becomes zero.

### Task 4.5

Verify the singular configuration numerically.

Keep the pan angle fixed at

$$
q_1=30^\circ
$$

and evaluate the Jacobian for several different tilt angles, including the singular configurations identified in Task 4.4.

For each configuration, calculate the rank of the Jacobian using

```python
np.linalg.matrix_rank()

````python
# Task 4.5 - Solution

q1 = np.deg2rad(30)

tilt_angles = [
    -90,
    -60,
     0,
     60,
     90
]

for angle in tilt_angles:

    q2 = np.deg2rad(angle)

    J = camera_jacobian(q1, q2)

    rank = np.linalg.matrix_rank(J)

    print(
        f"q2 = {angle:>6.1f} degrees | "
        f"rank(J) = {rank}"
    )
````
