#### Create row and column vectors
Check whether you create a row or a column vector using the following vector definition.

````python
# Creating a vector (i.e. 1 row with n values in columns)
vector = np.array([1, 2, 3]).reshape(3,1)

print("Vector: {}".format(vector))
print("Vector shape: {}".format(vector.shape))
````

It is possible to be lazy with vectors and merely use ```np.array([1, 2, 3])```, but the shape may not be correctly expressed.  They are expressed as 1D arrays with a length but with no defined rows or columns.  In textbooks vectors are written vertically, i.e. more rows than columns, for example $\textbf{v}=\begin{bmatrix}x \\ y \\ z \\ \end{bmatrix}$, which has shape $(3,1)$.  Matrices by comparison must be given a shape to make sense, however, a shapeless vector can be "reshaped" into a matrix if necessary.  Most of the time, NumPy will be clever enough to figure out what shape a vector _should_ be to perform an operation.

In most cases this won't cause an issue, but always remember that the lack of shape may trip you up!  Reduce the possibilities of headscratching, tears, and swearing at your code in the future by specifying the shape when things are getting more complicated, this also allows you to follow maths exactly as written in a textbook or paper.

It may be common to see something like $\textbf{v}=\begin{bmatrix} x & y & z \end{bmatrix}^{T}$, which means, it has been written "horizontally", but the transpose $^{(T)}$ converts it to vertical.  This is often exploited in academic papers where text space is a premium.

````python
lazy = np.array( [1,0,0] )
explicit = np.array( [1,0,0] ).reshape(3,1) # Reshape into a vertical array (as a vector would be written in a textbook)

print('lazy vector shape: {}'.format(lazy.shape))
print('lazy transposed shape: {}'.format(np.transpose(lazy).shape))
print('explicit vector shape: {}'.format(explicit.shape))
print('explicit transposed shape: {}'.format(np.transpose(explicit).shape))
````

### Task 3.3 

In this task, you will explore several important **matrix operations** using **NumPy**:

1. Matrix Transpose  
   - Flip a matrix over its diagonal (rows become columns and vice versa).  
   - This can be done using `np.transpose(matrix)` or the shorthand `matrix.T`.  

2. Matrix Multiplication (Dot Product) 
   - Multiply the original square matrix with its transpose using `np.dot()`.  
   - Observe how the result differs from element-wise multiplication.  

3. Determinant of a Matrix  
   - Calculate the determinant of a square matrix using `np.linalg.det()`.  
   - The determinant is a scalar value that provides important information about the matrix (e.g., whether it is invertible).  

4. Inverse of a Matrix  
   - Find the inverse of a square matrix using `np.linalg.inv()`.  
   - The inverse is the matrix that, when multiplied by the original, gives the identity matrix.

````python
# Task 3.3.1 
# Transpose - can also use .T after a variable as a shorthand, e.g. matrix.T == np.transpose(matrix)
square_matrix = np.array([[1,2], [3,4]])
############################################################
# Enter your codes
transpose_matrix = square_matrix.T
############################################################

print("\nOriginal Matrix: \n{}".format(square_matrix))
print("Transposed Matrix: \n{}".format(transpose_matrix))

# Task 3.3.2 
# Matrix Dot Product (Multiplication) of square_matrix and its transpose
############################################################
# Enter your codes
result_matrix_mult = square_matrix.dot(transpose_matrix)
############################################################

print("\nMatrix Multiplication:")
print("'Original Matrix' DOT 'Transposed Matrix':\n{}".format(result_matrix_mult))

# Task 3.3.3 Determinant of square_matrix
square_matrix = np.array([[1,2], [3,4]])
############################################################
# Enter your codes
determinant = np.linalg.det(square_matrix)
############################################################

print("\nDeterminant:\n{:.3}".format(determinant)) # Shorten output to 3 figures or less using ":.3"

# Task 3.3.4 Inverse of square_matrix
############################################################
# Enter your codes
inverse_matrix = np.linalg.inv(square_matrix)
############################################################

print("\nInverse Matrix: \n{}".format(inverse_matrix))
````

### Task 3.4 Equivalence check
You can also check if two matrices are equivalent with ```np.allclose```, which checks to see if each element is close enough within a tolerance.  This is more robust than a simple logical check (e.g. ```x==y```), especially if there have been some rounding/floating point errors which have snuck in.

````python
# Check if two matrices are the same
# The identity matrix should be equivalent when transposed (i.e. symmetric)
print(np.allclose(identity_matrix, identity_matrix.T))

# The square matrix should return False
print(np.allclose(square_matrix, square_matrix.T))
````
